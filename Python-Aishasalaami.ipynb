```python
#Automated transaction tier and discount assignmennt
order_value = 750.00
tier = ""
discount = 0.0
# 1. Branching Decision Ladder (evaluated top-to-bottom)
if order_value >= 1000.00:
  tier = "Platinum Enterprice"
  discount = 0.20
elif order_value >= 500.00:
    tier = "Gold preferred"
    discount = 0.10
elif order_value >= 150.00:
      tier = "Silver Preferred"
      discount = 0.05
else:
        tier = "Basic Retail"
        discount = 0.00
print()

# 2. Execution of Business Logic
net_billed = order_value * (1-discount)
print(f"Customer Tier: {tier} | Discount Rate: {discount*100:.0f}%")
print(f"Gross: ${order_value:.2f} | Net Billed: ${net_billed:.2f}")
```

```python
#1. Defining a custom function
def calculate_net_revenue(units_sold, unit_price, discount_rate=0.0):
    """
    Calculates total gross and net revenue after discount deduction.

    Parameters:
    units_sold (int): Number of units purchased.
    unit_price (float): List catalogue selling price.
    discount_rate (float): Applied discount (default 0.0).

    Returns:
    float: Net revenue after discount.
    """
    # Input validation guard
    if units_sold <= 0 or unit_price < 0:
        return 0.0

    subtotal = units_sold * unit_price
    net_revenue = subtotal * (1 - discount_rate)

    return round(net_revenue, 2)


# 2. Invoking (calling) the function
order_1 = calculate_net_revenue(
    units_sold=10,
    unit_price=49.99,
    discount_rate=0.10
)

order_2 = calculate_net_revenue(
    units_sold=4,
    unit_price=129.50
)  # Uses default discount

print(f"Order 1 Net: ${order_1:.2f} | Order 2 Net: ${order_2:.2f}")
```
